```python
from datasets import load_dataset

# 加载完整数据集
dataset = load_dataset("json", data_files={"data": "../dataset/shanghai/shanghai_hf_data.jsonl"})
full_dataset = dataset['data']

# 查看加载成功后的数据集
print(full_dataset)
```

```python
# 在加载的数据集上打乱 顺便设定随机种子
shuffled_dataset = full_dataset.shuffle(seed=42)  # seed=42 是为了结果可复现
```

```python
# 按 80:20 进行分割
train_test_split = shuffled_dataset.train_test_split(test_size=0.2)  

# 获取训练集和测试集
train_dataset = train_test_split["train"]
test_dataset = train_test_split["test"]

print(f"训练数据数量: {len(train_dataset)}")
print(f"测试数据数量: {len(test_dataset)}")
```

```python
train_dataset.save_to_disk("../dataset/shanghai/train")
test_dataset.save_to_disk("../dataset/shanghai/test")
```

```python
from datasets import load_from_disk

# 从磁盘加载
train_dataset = load_from_disk("../dataset/shanghai/train")
test_dataset = load_from_disk("../dataset/shanghai/test")

print(train_dataset)
print(test_dataset)
```
